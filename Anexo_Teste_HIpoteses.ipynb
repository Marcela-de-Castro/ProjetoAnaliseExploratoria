Adotou-se nível de significância $\alpha = 0,05$ para todos os testes formais de hipótese.

Pergunta 11: Teste A/B sobre o Impacto das Promoções (Desconto)

$H_0$: $\mu_{\text{Desconto}} = \mu_{\text{Sem Desconto}}$ (O valor médio de venda do grupo com desconto é igual ao grupo sem desconto).

$H_1$: $\mu_{\text{Desconto}} \neq \mu_{\text{Sem Desconto}}$ (O valor médio de venda diverge entre os grupos).

Teste Aplicado: Teste t de Welch para duas amostras independentes.

Resultados: Estatística $t = -15,738$, $p\text{-valor} = 5,82 \times 10^{-55}$ ($df = 7758,5$).

Conclusão de Negócio: Rejeita-se $H_0$. Há diferença estatisticamente significativa. O grupo com desconto apresentou valor médio de venda significativamente menor (R$ 205,93 vs R$ 256,41). A política de descontos atual está reduzindo a margem e a receita por ticket sem compensar em quantidade de unidades vendidas.

Pergunta 12: Teste A/B - Semana da Black Friday vs. Semana do Natal

$H_0$: $\mu_{\text{Diária BF}} = \mu_{\text{Diária Natal}}$ (A média de vendas diárias na semana da Black Friday é igual à da semana do Natal).

$H_1$: $\mu_{\text{Diária BF}} \neq \mu_{\text{Diária Natal}}$ (A média de vendas diárias difere entre as duas semanas).

Teste Aplicado: Teste t de Student para duas amostras independentes ($n = 7$ dias por grupo).

Resultados: Estatística $t = -1,524$, $p\text{-valor} = 0,1534$ ($df = 12$).

Intervalos de Confiança (95%):

Média Diária Black Friday: R$ 10.921,73 [IC 95%: R$ 9.019,70 a R$ 12.823,75]Média Diária Natal: R$ 12.387,75 [IC 95%: R$ 11.000,94 a R$ 13.774,55]

Conclusão de Negócio: Não se rejeita $H_0$ ao nível de 5%. Embora o Natal tenha apresentado faturamento diário médio absoluto superior (R$ 12.387 vs R$ 10.921), a variação não atinge significância estatística devido ao tamanho amostral diário ($n=7$) e à oscilação natural das semanas.

Pergunta 14: Teste A/B - Mês da Black Friday (Novembro) vs. Mês Anterior (Outubro)

$H_0$: $\mu_{\text{Novembro}} = \mu_{\text{Outubro}}$ (O valor médio de vendas no mês da Black Friday é igual ao mês de Outubro).
$H_1$: $\mu_{\text{Novembro}} \neq \mu_{\text{Outubro}}$ (O valor médio de vendas difere entre Novembro e Outubro).

Teste Aplicado: Teste t de Welch para duas amostras independentes.

Resultados: Estatística $t = -1,500$, $p\text{-valor} = 0,1337$ ($df = 3351,4$).

Conclusão de Negócio: Não se rejeita $H_0$. Não há evidência estatística de que o mês da Black Friday tenha gerado aumento de vendas em relação a Outubro. As campanhas de Black Friday não conseguiram alavancar a receita total do mês de Novembro.
