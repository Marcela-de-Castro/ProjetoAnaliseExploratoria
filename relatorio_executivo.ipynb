Relatório Executivo e Estratégico (Equipe de Marketing e Diretoria)Diagnóstico e Resumo Executivo
Analisamos 20.075 transações de vendas do último ano de operação com o objetivo de otimizar a receita, compreender o comportamento do consumidor e testar rigorosamente as hipóteses de negócio da rede.

Priorização de Categorias: A categoria Alimentos Frescos lidera isoladamente com R$ 1.365.309,00 (27,6% da receita total), seguida por Produtos de Mercearia (R$ 529.653,60) e Laticínios e Ovos (R$ 459.960,50).

Eficácia Promocional Ineficiente: O desconto médio praticado reduziu o ticket médio de R$ 256,41 (sem desconto) para R$ 205,93 (com desconto). Estatisticamente, as promoções não alavancaram o volume vendido a ponto de compensar a redução de margem, gerando uma retração na receita por transação.

Comportamento por Faixa Etária: O público Adulto (36-59 anos) é o maior gerador de receita (42,6% do total), seguido por Jovens (31,1%) e Idosos (26,3%). Contudo, o ticket médio por compra é virtualmente idêntico em todas as idades (~R$ 242 a R$ 247).

Mitos de Sazonalidade (Black Friday vs. Natal): A análise inferencial demonstrou que as vendas da semana da Black Friday e do mês de Novembro não apresentaram aumento estatisticamente significativo em comparação a meses convencionais ou ao período natalino.
